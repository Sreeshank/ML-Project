# Phase 5 — validated prediction
Run `python -m src.predict prepare` from the project root first. Predictions are estimates, not experimentally verified efficiencies.

````
from pathlib import Path
import sys
import json
root = Path.cwd()
if root.name == 'notebooks':
    root = root.parent
sys.path.insert(0, str(root))
from src.predict import load_bundle, predict_records
pipeline, metadata = load_bundle(root / 'models/app_bundle')
inputs = json.loads((root / 'models/app_bundle/example_input.json').read_text())
result = predict_records(pipeline, metadata, inputs, explain=True)
print(json.dumps(result, indent=2))
````

Launch the local interface from the project root:
```bash
python -m streamlit run app.py --server.port 8501
```
The model base value plus all grouped contributions reconstructs the prediction. Contributions describe associations, not causal effects.
