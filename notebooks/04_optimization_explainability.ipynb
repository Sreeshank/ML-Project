# Phase 4: validation tuning and explanations
The optimized model has no new test score. Explanations describe predictive associations, not causal effects.

```
from pathlib import Path
import json
import pandas as pd
from IPython.display import display, Image
root = Path.cwd()
if root.name == 'notebooks':
    root = root.parent
run = root / 'reports/phase4_run_001'
explanations = root / 'reports/phase4_explanations_001'
assert (run / 'COMPLETE').exists() and (explanations / 'COMPLETE').exists()
display(pd.read_csv(run / 'validation_comparison.csv').sort_values('RMSE'))
display(json.loads((run / 'selection.json').read_text()))
```

```
display(json.loads((explanations / 'explanation_metadata.json').read_text()))
for filename in ['shap_global.png', 'shap_summary.png', 'shap_local.png', 'permutation_importance.png']:
    display(Image(filename=str(explanations / 'figures' / filename)))
display(pd.read_csv(explanations / 'shap_local_raw.csv'))
display(pd.read_csv(explanations / 'permutation_importance.csv').head(15))
```

Raw-source SHAP groups sum encoded and engineered contributions. Permutation changes raw fields and recomputes descendants. Correlations and implausible permutations limit causal interpretation.
